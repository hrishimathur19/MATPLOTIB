```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
x = np.linspace(1,5,11)
y = np.linspace(1,5,11)
```

```python
fig = plt.figure(figsize=(12,8), dpi=800)
axis1 = fig.add_axes([0.1,0.1,1,1])
axis1.plot(x,y)
axis2 = fig.add_axes([0.4,0.4,0.4,0.4])
axis2.plot(x,y)
axis3 = fig.add_axes([0.2,0.2,0.2,0.2])
axis3.plot(x,y)
```
